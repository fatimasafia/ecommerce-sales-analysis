```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the CSV file
df = pd.read_excel("Online Retail Data Set.xlsx")

# Check the data
print(df.shape)
print(df.head())
print(df.info())
print(df.describe())   # Statistics
```

```python
# 1. Check missing values
print(df.isnull().sum())

# 2. Remove rows with missing CustomerID (we can't analyze customers without ID)
df = df.dropna(subset=['CustomerID'])

# 3. Remove rows with missing Description
df = df.dropna(subset=['Description'])

# 4. Remove cancelled orders (negative quantities)
df = df[df['Quantity'] > 0]

# 5. Remove rows with zero or negative price
df = df[df['UnitPrice'] > 0]

# 6. Create a Revenue column
df['Revenue'] = df['Quantity'] * df['UnitPrice']

# 7. Extract Year-Month from InvoiceDate
df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'])
df['YearMonth'] = df['InvoiceDate'].dt.to_period('M').astype(str)

# 8. Convert CustomerID to integer (not float)
df['CustomerID'] = df['CustomerID'].astype(int)

# Check final shape
print(f"Cleaned dataset shape: {df.shape}")
print(df.head())

```

```python
import pandas as pd

df = pd.read_csv("Online Retail Data Set.csv", encoding="latin1")

print(df.shape)
```

```python
print(df.shape)
print(df.columns)
print(df.info())
print(df.isnull().sum())
print(df.duplicated().sum())
print(df.describe())
```

```python
# Remove duplicate rows
df = df.drop_duplicates()

# Remove rows with missing CustomerID
df = df.dropna(subset=["CustomerID"])

# Remove rows with missing Description
df = df.dropna(subset=["Description"])

# Keep only positive quantities
df = df[df["Quantity"] > 0]

# Keep only positive prices
df = df[df["UnitPrice"] > 0]

# Create Revenue column
df["Revenue"] = df["Quantity"] * df["UnitPrice"]

# Convert InvoiceDate correctly
df["InvoiceDate"] = pd.to_datetime(
    df["InvoiceDate"],
    format="mixed",
    dayfirst=True
)

# Create Year-Month column
df["YearMonth"] = df["InvoiceDate"].dt.to_period("M").astype(str)

# Convert CustomerID to integer
df["CustomerID"] = df["CustomerID"].astype(int)

# Check result
print("Cleaned dataset shape:", df.shape)
print(df.head())
```

```python
import os

os.makedirs("data", exist_ok=True)
df.to_csv("data/cleaned_sales.csv", index=False)
print(os.listdir("data"))
```

```python
import os

# Create folder for saved charts
os.makedirs("images", exist_ok=True)

# QUESTION 1: Monthly Revenue Trend
monthly_revenue = df.groupby('YearMonth')['Revenue'].sum().reset_index()

plt.figure(figsize=(12,5))
sns.barplot(data=monthly_revenue, x='YearMonth', y='Revenue')
plt.title('Monthly Revenue Trend')
plt.xticks(rotation=45)
plt.tight_layout()

# Save the chart
plt.savefig('images/monthly_revenue.png')

# Display the chart
plt.show()
```

```python
# QUESTION 2: Top 10 Products by Revenue
top_products = df.groupby('Description')['Revenue'].sum().sort_values(ascending=False).head(10)

plt.figure(figsize=(10,6))
top_products.plot(kind='barh')
plt.title('Top 10 Products by Revenue')
plt.tight_layout()
plt.savefig('images/top_products.png')
plt.show()
```

```python
# QUESTION 3: Top 10 Countries by Revenue
top_countries = df.groupby('Country')['Revenue'].sum().sort_values(ascending=False).head(10)

plt.figure(figsize=(10,6))
top_countries.plot(kind='bar')
plt.title('Top 10 Countries by Revenue')
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig('images/top_countries.png')
plt.show()
```

```python
# QUESTION 4: Key Metrics (print these for your README)
print(f"Total Revenue: ${df['Revenue'].sum():,.2f}")
print(f"Total Orders: {df['InvoiceNo'].nunique()}")
print(f"Total Customers: {df['CustomerID'].nunique()}")
print(f"Average Order Value: ${df.groupby('InvoiceNo')['Revenue'].sum().mean():.2f}")
```
